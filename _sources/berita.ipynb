 # word embedding

```python
import pandas as pd

# Baca hasil preprocessing ringkasan
df = pd.read_csv("hasil_preprocessing_ringkasan.csv")

# Ambil kolom processed sebagai corpus
corpus = df["processed"].astype(str).tolist()

print("Jumlah dokumen:", len(corpus))
print(corpus[:3])  # preview 3 dokumen pertama
```

```python
from nltk.tokenize import word_tokenize
import ast

# Tokenisasi tiap dokumen
tokenized_corpus = [ast.literal_eval(doc) for doc in corpus]

# Cek hasil tokenization
print("Contoh tokenization dokumen pertama:")
print(tokenized_corpus[0][:20])
```

```python
from gensim.models import Word2Vec

# Training model Word2Vec (CBOW -> sg=0)
model_cbow = Word2Vec(
    sentences=tokenized_corpus,
    vector_size=100,  # ukuran embedding
    window=5,
    min_count=2,      # kata yang muncul <2 kali akan diabaikan
    sg=0,             # 0 = CBOW, 1 = Skip-gram
    workers=4
)

# Simpan model
model_cbow.save("cbow_ringkasan.model")

print("Model CBOW berhasil dilatih dan disimpan!")
```

```python
# Ambil embedding untuk kata tertentu
print("Vector untuk kata 'indonesia':")
print(model_cbow.wv['indonesia'])

# Cek kata yang mirip
print("\nKata mirip 'ekonomi':")
print(model_cbow.wv.most_similar('ekonomi'))
```

 # perbandingan TF-IDF dan Word embedding

```python
# Load data hasil preprocessing
df = pd.read_csv("hasil_preprocessing_ringkasan.csv")

# Pastikan ada label untuk klasifikasi
# Ambil text & label
X_text = df["processed"].astype(str)
y = df["Kategori"].astype(str)

print("Jumlah data:", len(df))
print(df[["Judul", "Kategori"]].head())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# TF-IDF Vectorizer
tfidf = TfidfVectorizer()
X_tfidf = tfidf.fit_transform(X_text)

print("Shape TF-IDF:", X_tfidf.shape)
```

```python
from nltk.tokenize import word_tokenize
from gensim.models import Word2Vec
import numpy as np

# Tokenisasi
tokenized = [word_tokenize(doc) for doc in X_text]

# Training model Word2Vec CBOW
model_cbow = Word2Vec(
    sentences=tokenized,
    vector_size=100,
    window=5,
    min_count=2,
    sg=0,   # CBOW
    workers=4
)

# Fungsi untuk dapatkan rata-rata embedding per dokumen
def get_avg_vector(tokens, model, vector_size=100):
    vectors = [model.wv[word] for word in tokens if word in model.wv]
    if len(vectors) > 0:
        return np.mean(vectors, axis=0)
    else:
        return np.zeros(vector_size)

X_w2v = np.array([get_avg_vector(tokens, model_cbow, 100) for tokens in tokenized])

print("Shape Word2Vec:", X_w2v.shape)
```

```python
from sklearn.model_selection import train_test_split

# Bagi data train/test
X_train_tfidf, X_test_tfidf, y_train, y_test = train_test_split(
    X_tfidf, y, test_size=0.2, random_state=42
)

X_train_w2v, X_test_w2v, _, _ = train_test_split(
    X_w2v, y, test_size=0.2, random_state=42
)

print("Train size:", len(y_train))
print("Test size :", len(y_test))
```

```python
from sklearn.model_selection import train_test_split

# Bagi data train/test
X_train_tfidf, X_test_tfidf, y_train, y_test = train_test_split(
    X_tfidf, y, test_size=0.2, random_state=42
)

X_train_w2v, X_test_w2v, _, _ = train_test_split(
    X_w2v, y, test_size=0.2, random_state=42
)

print("Train size:", len(y_train))
print("Test size :", len(y_test))
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score

# Softmax untuk TF-IDF
clf_tfidf = LogisticRegression(max_iter=1000)
clf_tfidf.fit(X_train_tfidf, y_train)
y_pred_tfidf = clf_tfidf.predict(X_test_tfidf)

# Softmax untuk Word2Vec
clf_w2v = LogisticRegression(max_iter=1000)
clf_w2v.fit(X_train_w2v, y_train)
y_pred_w2v = clf_w2v.predict(X_test_w2v)

print("=== Hasil Klasifikasi TF-IDF ===")
print("Akurasi:", accuracy_score(y_test, y_pred_tfidf))
print(classification_report(y_test, y_pred_tfidf, zero_division=0))

print("\n=== Hasil Klasifikasi Word2Vec (CBOW) ===")
print("Akurasi:", accuracy_score(y_test, y_pred_w2v))
print(classification_report(y_test, y_pred_w2v, zero_division=0))
```
