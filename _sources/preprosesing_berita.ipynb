```python
import pandas as pd
import string
import re
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from spellchecker import SpellChecker
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
import nltk
```

```python
# Baca file
df = pd.read_csv("data_artikel_berita.csv")
col_text = "artikel" if "artikel" in df.columns else df.columns[0]
```

```python
# Stopword Bahasa Indonesia
stop_words = set(stopwords.words('indonesian'))

# Spell checker (pakai pyspellchecker)
spell = SpellChecker(language=None)
spell.word_frequency.load_words(["indonesia", "pemerintah", "ekonomi", "politik"])

# Stemming
factory = StemmerFactory()
stemmer = factory.create_stemmer()
```

```python
def preprocess_text(text):
    if not isinstance(text, str):
        return ""
    text = text.lower()
    tokens = word_tokenize(text)
    tokens = [w for w in tokens if w not in stop_words]
    tokens = [re.sub(f"[{re.escape(string.punctuation)}]", "", w) for w in tokens]
    tokens = [w for w in tokens if w.strip() != ""]
    tokens = [spell.correction(w) if spell.correction(w) else w for w in tokens]
    tokens = [stemmer.stem(w) for w in tokens]
    return " ".join(tokens)
```

```python
text_columns = df.select_dtypes(include=["object"]).columns
for col in text_columns:
    df[f"{col}_processed"] = df[col].astype(str).apply(preprocess_text)
```

```python
df.to_csv("hasil.csv", index=False)
```
