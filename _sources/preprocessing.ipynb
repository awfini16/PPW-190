 # Berita

```python
import pandas as pd
import string
import re
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
import nltk
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
# Baca file
df = pd.read_csv("berita.csv")
col_text = "artikel" if "artikel" in df.columns else df.columns[0]
df.head()
```

```python
# Case Folding: ubah semua huruf jadi lowercase
df["case_folding"] = df["Ringkasan"].str.lower()

print("=== Hasil Case Folding ===")
print(df[["Ringkasan", "case_folding"]].head())
```

```python
# Cleaning: hapus angka, tanda baca, simbol
df["cleaning"] = df["case_folding"].apply(lambda x: re.sub(r"[^a-zA-Z\s]", "", x))

print("=== Hasil Cleaning ===")
print(df[["case_folding", "cleaning"]].head())
```

```python
# Tokenizing dengan NLTK
df["tokenizing"] = df["cleaning"].apply(lambda x: word_tokenize(x))

print("=== Hasil Tokenizing ===")
print(df[["cleaning", "tokenizing"]].head())
```

```python
# Stopword Removal menggunakan NLTK (Bahasa Indonesia)
stop_words = set(stopwords.words("indonesian"))

df["stopword_removal"] = df["tokenizing"].apply(lambda tokens: [w for w in tokens if w not in stop_words])

print("=== Hasil Stopword Removal ===")
print(df[["tokenizing", "stopword_removal"]].head())
```

```python
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
stemmer = StemmerFactory().create_stemmer()

# Stemming
df["stemming"] = df["stopword_removal"].apply(lambda tokens: [stemmer.stem(w) for w in tokens])

print("=== Hasil Stemming ===")
print(df[["stopword_removal", "stemming"]].head())
```

```python
# Processed = sama dengan stemming (tanpa join ulang)
df["processed"] = df["stemming"]

print("=== Hasil Akhir Preprocessing ===")
print(df[["Ringkasan", "processed"]].head())

# Simpan ke CSV lengkap (semua kolom tetap ada)
df.to_csv("hasil_preprocessing_ringkasan.csv", index=False)
```

# PTA

```python
# 1. Baca dataset
df = pd.read_csv("PPW_HasilCrawling_FakultasTeknik_partial.csv")
```

```python
print("===== Data Asli =====")
print(df.head())

text_column = df.columns[0]
```

```python
# 2. Case Folding + Cleaning
def clean_text(text):
    if pd.isna(text):
        return ""
    text = text.lower()                               
    text = re.sub(r"http\S+|www\S+|https\S+", '', text) 
    text = re.sub(r"\d+", "", text)                   
    text = text.translate(str.maketrans("", "", string.punctuation)) 
    text = re.sub(r"\s+", " ", text).strip()          
    return text

df["clean_text"] = df.iloc[:,0].astype(str).apply(clean_text)  # ambil dari kolom pertama
```

```python
# 3. Tokenisasi
df["tokens"] = df["clean_text"].apply(word_tokenize)
```

```python
# 4. Stopword Removal
stop_words = set(stopwords.words("indonesian"))
df["tokens_nostop"] = df["tokens"].apply(lambda x: [w for w in x if w not in stop_words])
```

```python
# 5. Stemming
factory = StemmerFactory()
stemmer = factory.create_stemmer()
df["stemmed"] = df["tokens_nostop"].apply(lambda x: [stemmer.stem(w) for w in x])
```

```python
# 6. Gabungkan kembali
df["processed_text"] = df["stemmed"].apply(lambda x: " ".join(x))
```

```python
print("\n===== Contoh Hasil Preprocessing =====")
print(df[[text_column, "clean_text", "tokens", "tokens_nostop", "processed_text"]].head())

df.to_csv("hasil_preprocessing_PTA.csv", index=False)
```
